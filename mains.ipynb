```python
def read_students()->list[dict[str,str|int]]:
    """Read and validate each line in ``students.txt``.

    Blank lines are skipped. Records that do not have exactly five fields,
    that contain an empty score, that contain a non-numeric score, or that
    have any score outside  0–100 are written to
    ``errors.log`` and excluded from the returned list.

    Returns:
        A list of dictionaries, each with keys ``student_id``, ``name``,
        ``python_score``, ``sql``, and ``excel``. The list is empty if the
        file cannot be read or contains no valid records.
    """
    students_data = []
    try:
        with open('students.txt', 'r') as file, open('errors.log', 'w') as bad_file:

            for line in file:
                line = line.strip()

                if line == '':
                    continue
                
                fields = line.split(",")
                if len(fields) == 5:
                    if fields[0]==''or fields[1]=='' or fields[2] == '' or fields[3] == '' or fields[4] == '':
                        bad_file.write(line + "\n")
                        # print("Empty field")
                        continue
                    try:    
                        student_id = fields[0]
                        name = fields[1]
                        python_score = int(fields[2])
                        sql = int(fields[3])
                        excel = int(fields[4])
                    except:
                        bad_file.write(line + '\n')
                        continue

                    if all(0 <= val <= 100 for val in [python_score, sql, excel]):
                        # print('Good')
                        students_data.append({
                            'student_id' : student_id,
                            'name': name,
                            'python_score': python_score,
                            'sql': sql,
                            'excel': excel
                        })
                    else:
                        bad_file.write(line + "\n")
                        #print(f"Out of range scores: {line}")
                else:
                    bad_file.write(line + "\n")
                    # print(f"invalid format:{line}")
                    
    except Exception as e:
        print(f"Error: {e}")

    return students_data
```

```python
data = read_students()
print(data)
```

```python
def calculate_average(scores:list)->list[dict[str,str|int]]:
    """Attach the rounded average of the three subject scores to each record.

    Args:
        scores: A list of validated student records, each containing the
            keys ``student_id``, ``name``, ``python_score``, ``sql``, and
            ``excel``.

    Returns:
        A new list of dictionaries identical to the input, with an extra
        ``average`` key holding the mean of the three subjects, rounded to
        two decimal places.
    """
    name_list = []
    for score in scores:
        student_id = score['student_id']
        name = score['name']
        python_score=score["python_score"]
        sql = score['sql']
        excel = score['excel']

        average = round((python_score + sql + excel)/3,2)

        name_list.append({
            'student_id' : student_id,
            'name' : name,
            'python_score' : python_score,
            'sql' : sql,
            'excel' : excel,
            'average' : average
        })
    return name_list
```

```python
average_list=calculate_average(data)
print(average_list)
```

```python

def grade_score(scores:list)->list[dict[str,str]]:
    
    grades=[]
    for score in scores:
        student_id = score['student_id']
        python_score=score["python_score"]
        sql = score['sql']
        excel = score['excel']
        name = score['name']
        average = score['average']


        def get_grades(args:int)->str:
            if 70 <= args <= 100 :
                return"A"
                
            elif 60 <= args <= 69 :
                return"B"
                
            elif 50 <= args <= 59 :
                return'C'
            elif 40 <= args <= 49 :
                return'D'

        def get_category(cat:int)-> str:
            if 70 <= cat <= 100 :
                return"Excellent"
                
            elif 60 <= cat <= 69 :
                return"Good"
                
            elif 50 <= cat <= 59 :
                return'Average'
            elif 40 <= cat <= 49 :
                return'Needs improvement'

        grades.append({
            'student_id' : student_id,
            'name' : name,
            'python_score' : get_grades(python_score) ,
            'sql' : get_grades(sql),
            'excel' : get_grades(excel),
            'average' : get_category(average)
        })

    return grades
        

```

```python
grades_list=grade_score(average_list)
print(grades_list)
```

```python
def generate_report(datas:list,average_list:list)->None:
    """Write a student report line for each pair of graded/averaged data.

    The two input lists are expected to be in the same order (as produced by
    :func:`grade_score` and :func:`calculate_average` respectively) so that
    ``zip`` pairs each student's grades with their raw scores.

    Args:
        datas: Graded records from :func:`grade_score`.
        average_list: Averaged records from :func:`calculate_average`.

    Returns:
        None. Writes the report to ``results.txt`` in write mode, truncating
        any previous report so it reflects the current data.
    """
    
    with open('results.txt','w') as file:
        for grade_data,score_data in zip(datas,average_list):
            student_id=grade_data['student_id']
            python_score=grade_data["python_score"]
            sql = grade_data['sql']
            excel = grade_data['excel']
            name = grade_data['name']
            average = grade_data['average']
            file.write(f"{name} with Student ID {student_id},got {score_data['sql']}:({sql})in SQL,{score_data['excel']}:({excel}) in Excel and {score_data['python_score']}:({python_score}) in Python. A total average of {score_data['average']} ({average})" + '\n')

        

    
```

```python
generate_report(grades_list,average_list)
```

```python
def average_info(averages:list)->float:
    """Append the class summary (highest, lowest, class average) to the report.

    Args:
        averages: Averaged records from :func:`calculate_average`.

   
    """
    with open('results.txt','a') as file:
        
        avg = []
        for average in averages:
            avg.append(average['average'])
        min_avg = min(avg)
        max_avg = max(avg)
        class_avg = round(sum(avg)/len(avg),2)


        file.write(f"The highest average is {max_avg},Lowest {min_avg},and the class average is {class_avg}")
        

```

```python
average_info(average_list)
```

```python
def find_student(students:list,grades:list):
    """Look up a student by ID and print their full result to the console.

    Args:
        students: Validated student records from :func:`read_students`.
        grades: Graded records from :func:`grade_score`, aligned by
            ``student_id``.

    Returns:
        None. Prints the student's details, or a message if the
        ID is empty or not found.
    """
    student_lookup = {student["student_id"]:student for student in students }
    grades_lookup = {grade["student_id"]: grade for grade in grades}
    user_input =''.join(input("Enter Student Id(E.g ST001):  ").strip().upper())
    if not user_input:
        print("Invalid Entry,Enter A Valid Student_id")
        return

    student_info = student_lookup.get(user_input)
    grade_info = grades_lookup.get(user_input)

    if not student_info or not grade_info:
        print("Invalid Entry,Enter A Valid Student_id")
        return
    try:
        if student_info and grade_info:
            print(f"Student_id: {student_info["student_id"]}")
            print(f"Name: {student_info['name']}")
            print(f"SQL Score: {student_info["sql"]}")
            print(f"SQL Grade: {grade_info["sql"]}")
            print(f"Python Score: {student_info["python_score"]}")
            print(f"Python Grade: {grade_info["python_score"]}")
            print(f"Excel Score: {student_info["excel"]}")
            print(f"Excel Grade: {grade_info["excel"]}")
            print(f"Your Average is {student_info["average"]}:'{grade_info['average']}' ")

    except Exception as e:
        print(e)
        


    
```

```python
def display_menu() -> None:
    """Print the main menu options to the console.

    Returns:
        None.
    """
    
    print(" STUDENT RESULTS PROCESSING SYSTEM")
    print(" 1. Load & validate students.txt")
    print(" 2. View class averages summary")
    print(" 3. Generate results report (results.txt)")
    print(" 4. Search student by ID")
    print(" 5. Add new student")
    print(" 6. Exit")
    


def main() -> None:
    """Run the Student Results Processing System once.

    Displays the menu, reads a single choice, and performs the corresponding
    action. All state lives in local variables; nothing is global.

    Returns:
        None.
    """
    # Local state shared across the single pass.
    students: list = []
    averages: list = []
    grades: list = []

    
    choice: str = input("Select option (1-6): ").strip()

    
    if choice == "1":
        try:
            students = read_students()
        except (FileNotFoundError, PermissionError) as exc:
            print(f"[ERROR] {exc}")
        else:
            if students:
                averages = calculate_average(students)
                grades = grade_score(averages)
                
            else:
                print("[WARN] No valid records were found.")
        finally:
            print("[INFO] Load operation complete.")

    elif choice == "2":
        if not averages:
            print("[WARN] No records loaded. Use option 1 first.")
        else:
            try:
                average_info(averages)
            except ValueError as e:
                print(f"[ERROR] {e}")
            except (PermissionError) as e:
                print(f"[ERROR] {e}")
            else:
                print("[OK] Summary written to results.txt.")

   
    elif choice == "3":
        if not students or not grades or not averages:
            print("[WARN] No records loaded. Use option 1 first.")
        else:
            try:
                generate_report(grades, averages)
            except (PermissionError) as exc:
                print(f"[ERROR] {exc}")
            else:
                print("[OK] Report written to results.txt.")

    
    elif choice == "4":
        if not students or not grades:
            print("[WARN] No records loaded. Use option 1 first.")
        else:
            try:
                find_student(students, grades)
            except ValueError as exc:
                print(f"[ERROR] {exc}")
            

    
    elif choice == "5":
        try:
            added = add_students(students)
        except (PermissionError) as exc:
            print(f"[ERROR] {exc}")
        else:
            if added:
                
                try:
                    students = read_students()
                    averages = calculate_average(students)
                    grades = grade_score(averages)
                except (FileNotFoundError, PermissionError) as exc:
                    print(f"[WARN] Could not refresh records: {exc}")
                finally:
                    print("[INFO] Refresh after add complete.")

    
    elif choice == "6":
        print("Goodbye.")

    else:
        print("[WARN] Invalid option. Please choose 1-6.")


```
